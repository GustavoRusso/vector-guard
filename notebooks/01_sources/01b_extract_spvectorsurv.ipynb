# Extract SPVECTORSURV vector detections (raw DwC-A)

Downloads the **latest** SPVECTORSURV Darwin Core Archive from the Spanish GBIF IPT and **overwrites** the local raw copy under `data/01_raw/vector_spvectorsurv/`.

- Source: [SPVECTORSURV](https://www.gbif.org/dataset/fc1dd9e3-ff53-4d57-9d8e-e29afbb955b8) (`spvectorsurv`)
- GBIF UUID: `fc1dd9e3-ff53-4d57-9d8e-e29afbb955b8`
- License: **CC BY-NC 4.0** (non-commercial; combined interim is multi-license)
- No GBIF account required
- Separate from VectorNet so `01_extract_vectors.ipynb` cannot wipe this directory

This notebook stops at raw validation (file list + occurrence smoke check). It does not clean or process the data. Cleaning / union with VectorNet is notebook `08`.

```python
from __future__ import annotations

import hashlib
import json
import shutil
import zipfile
from datetime import datetime, timezone
from pathlib import Path
from urllib.request import Request, urlopen

from tqdm import tqdm

from vector_guard.data import default_spvectorsurv_raw_dir, find_repo_root

REPO_ROOT = find_repo_root(Path.cwd())
RAW_DIR = default_spvectorsurv_raw_dir(REPO_ROOT)
ZIP_NAME = "spvectorsurv_dwca.zip"
ZIP_PATH = RAW_DIR / ZIP_NAME
MANIFEST_PATH = RAW_DIR / "manifest.json"

SOURCE_URL = "https://ipt.gbif.es/archive.do?r=spvectorsurv"
RESOURCE = "spvectorsurv"
GBIF_DATASET_UUID = "fc1dd9e3-ff53-4d57-9d8e-e29afbb955b8"
LICENSE = "CC BY-NC 4.0"
CITATION = (
    "Ministerio de Agricultura, Pesca y Alimentación. SPVECTORSURV: "
    "Monitoring competent vectors of Crimea Congo Haemorragic Fever (ticks), "
    "West Nile and Rift Valley Fever in Spain (mosquitos). "
    "Occurrence dataset (CC BY-NC 4.0). "
    "https://doi.org/10.15470/fn1bet"
)
DOWNLOAD_TIMEOUT_S = 300
CHUNK_SIZE = 1024 * 256

RAW_DIR, SOURCE_URL
```

## Wipe local raw directory

Every run replaces the previous local copy with the archive fetched now. Only `data/01_raw/vector_spvectorsurv/` is wiped (VectorNet raw is untouched).

```python
def wipe_raw_dir(path: Path) -> None:
    path.mkdir(parents=True, exist_ok=True)
    for child in path.iterdir():
        if child.name == ".gitkeep":
            continue
        if child.is_dir():
            shutil.rmtree(child)
        else:
            child.unlink()
    (path / ".gitkeep").touch(exist_ok=True)


wipe_raw_dir(RAW_DIR)
print(f"Wiped and recreated: {RAW_DIR}")
```

## Download latest DwC-A from Spanish IPT

```python
def download_file(url: str, dest: Path, timeout_s: int = DOWNLOAD_TIMEOUT_S) -> int:
    request = Request(url, headers={"User-Agent": "vector-guard/0.1"})
    with urlopen(request, timeout=timeout_s) as response:
        total = response.headers.get("Content-Length")
        total_bytes = int(total) if total and total.isdigit() else None
        written = 0
        with dest.open("wb") as out, tqdm(
            total=total_bytes,
            unit="B",
            unit_scale=True,
            unit_divisor=1024,
            desc=dest.name,
        ) as bar:
            while True:
                chunk = response.read(CHUNK_SIZE)
                if not chunk:
                    break
                out.write(chunk)
                written += len(chunk)
                bar.update(len(chunk))
    return written


zip_bytes = download_file(SOURCE_URL, ZIP_PATH)
print(f"Downloaded {zip_bytes:,} bytes -> {ZIP_PATH}")
```

## Extract archive and write manifest

```python
def sha256_file(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as f:
        for chunk in iter(lambda: f.read(CHUNK_SIZE), b""):
            digest.update(chunk)
    return digest.hexdigest()


with zipfile.ZipFile(ZIP_PATH) as zf:
    zf.extractall(RAW_DIR)

extracted = sorted(
    p.relative_to(RAW_DIR).as_posix()
    for p in RAW_DIR.rglob("*")
    if p.is_file() and p.name != ZIP_NAME
)

manifest = {
    "source_url": SOURCE_URL,
    "resource": RESOURCE,
    "gbif_dataset_uuid": GBIF_DATASET_UUID,
    "license": LICENSE,
    "fetched_at": datetime.now(timezone.utc).isoformat(),
    "zip_name": ZIP_NAME,
    "zip_bytes": ZIP_PATH.stat().st_size,
    "sha256": sha256_file(ZIP_PATH),
    "citation": CITATION,
    "extracted_files": extracted,
}
MANIFEST_PATH.write_text(json.dumps(manifest, indent=2) + "\n", encoding="utf-8")

print(f"Extracted {len(extracted)} files")
print(f"Wrote {MANIFEST_PATH}")
manifest
```
